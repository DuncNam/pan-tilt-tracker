```python
import sys
sys.executable

import numpy as np



#
# Recovery Model Training
#

# Function definitions

# Calculate the MSE
def fMSE(pred, label):
    return np.mean((pred-label)**2)

# Calculate the gradient
def fgradient(data,label,pred):
    xT = data.T
    n = len(label)
    return (2/n)*(xT @ (pred-label))

# Calculate the new weights
def fweight(w, gradient, lr):
    return w - (gradient * lr)

    

# Input the samples and features
x = np.array([[53,102,13.8],[56,76,13.4],[55,77,13.2],[56,70,13.4],[57,60,13.4],[55,91,13.2],[52,104,13.1]])

# Add a bias
ones = np.ones((x.shape[0], 1))
x = np.hstack((x,ones))
# print("Data: \n", x, "\n")

# Input the labels
y = np.array([96, 61, 64, 51, 37, 84, 94])
# print("Labels: \n", y, "\n")

# Input the initial weights
w = np.array([0.3,0.3,0.3,0.3])
# print("Initial weights: \n", w, "\n")

# Input a learning rate
lr = 0.00001
# print("Learning rate: \n", lr, "\n")

# Training Loop

epochs = 10000

for epoch in range(epochs):
    # Make prediction
    y_pred = x @ w

    # Find MSE
    MSE = fMSE(y_pred, y)

    # Nan check
    if np.isnan(MSE):
        print("Nan detected in MSE during Epoch ", epoch)
        break

    # Find gradient
    gradient = fgradient(x, y, y_pred)

    # Find new weights
    w = fweight(w, gradient, lr)

if epoch == epochs-1:
    print("After ", epochs, " epochs, the MSE is ", MSE, "\n")
    print("The final weights are ", w, "\n")
    print("The predicted values are ", y_pred, "\n")
```

```python
# Predict Scores Given Metrics
given = [80, 50, 12, 1]
predicted = given @ w
predicted
```
